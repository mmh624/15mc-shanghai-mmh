## Literature Review

### 1. The 15-Minute City: Origins and Core Framework

The 15-minute city (15MC) framework was first proposed by Carlos Moreno in 2016 and gained international traction following its adoption as the centrepiece of Paris Mayor Anne Hidalgo's 2020 re-election platform. The concept's canonical academic statement appears in Moreno et al. (2021), published in *Smart Cities*, which defines the framework around six essential urban functions — living, working, commerce, healthcare, education, and entertainment — all reachable within fifteen minutes on foot or by bicycle from any residential location. Moreno et al. argue that restructuring cities around this proximity principle simultaneously addresses carbon emissions, social isolation, and post-pandemic urban resilience. Crucially, the framework is grounded in *soft mobility*: walking and cycling are the privileged modes, and the car is explicitly excluded from the fifteen-minute calculation. This distinction matters for the present project because it defines the methodological baseline: proximity scores derived here use pedestrian and cycling isochrones only, with car-mode isochrones computed separately as a spatial inequality diagnostic.

The framework's six functions map closely onto the project's baseline scoring layer. Entertainment — the dimension this project foregrounds — is listed as one of the six universal needs, yet it receives less methodological attention than healthcare or education in existing empirical literature. This gap is the point of entry for Track B.

### 2. Equity Critique: The 15MC Is Not Socially Neutral

Pozoukidou and Chatziyiannaki (2021), writing in *Sustainability*, offer the most direct structural critique of the 15MC concept. Their cross-case analysis of three adopting cities argues that the framework commits a form of physical determinism: by treating walkability as a design-led solution, it sidesteps the socio-economic processes — rising land values, displacement, gentrification — that proximity-based planning tends to accelerate. Neighbourhoods that score highest on fifteen-minute access criteria are, almost by definition, dense, mixed-use, and centrally located. These are also the neighbourhoods most exposed to rent pressure and the displacement of lower-income residents who the framework nominally aims to serve.

For Shanghai specifically, this critique has a spatial translation. The Inner Ring Road encloses some of the highest land-value territory in mainland China. If nightlife access concentrates within the inner ring — as preliminary evidence and local knowledge suggest — then the 15MC entertainment score will map almost perfectly onto housing unaffordability. The project's spatial equity analysis (Sprint 4) directly operationalises this critique by comparing mean Track B scores across the four ring zones and computing the proportion of high-entertainment cells that also have post-23:00 metro coverage.

Pozoukidou and Chatziyiannaki also note that the concept "is not a radical new idea" but rather a repackaging of longstanding neighbourhood-unit planning principles. This observation is methodologically useful: it suggests that isochrone-based access scoring, the core analytical tool of this project, has a well-established theoretical grounding predating the 15MC branding.

### 3. Methodological Template: Scenario-Based Access Analysis

Limerick et al. (2023), published in *Urban Forestry & Urban Greening*, provide the closest methodological analogue to the present project. Their study examines fifteen-minute walking access to community gardens across New York City, using a grid-based analytical framework, spatial optimisation modelling, and scenario analysis to ask how and where urban amenities should be expanded to maximise equitable coverage. Their key finding — that over half of NYC residents already have walking access to a community garden, but that access concentrates in lower-income, higher-diversity neighbourhoods rather than the reverse — illustrates how proximity scores can reveal counter-intuitive equity patterns.

The methodological relevance is direct. Limerick et al. demonstrate: (a) the use of a fixed spatial grid as the analytical unit rather than administrative boundaries, which avoids the modifiable areal unit problem; (b) the operationalisation of "access" as binary within-isochrone presence rather than continuous distance decay, which simplifies interpretation and aligns with the 15MC's own logic; and (c) scenario modelling to assess policy intervention. This project adopts (a) and (b) directly in its 500-metre grid construction and binary indicator scoring approach. The scenario analysis element is deferred to potential future work.

One limitation of the Limerick et al. framework, noted here for transparency, is that it treats all walking access as equivalent regardless of time-of-day. This is adequate for community gardens but insufficient for nightlife, which is inherently time-contingent. The present project extends the methodology by introducing a temporal dimension: the `last_train_after_23` flag, which conditions the effective accessibility of nightlife venues on the availability of a return transit option after the peak entertainment window closes.

### 4. The Nightlife–Transit Gap: A Structurally Neglected Research Problem

McArthur, Robin, and Smeds (2019), in *Transportation Research Part A*, examine the spatiotemporal dimensions of transport equity specifically in the context of London's night-time economy. Their central finding is that London's night-time transport strategies — including the Night Tube launched in 2016 — were designed primarily to grow the nighttime economy as an economic sector, not to address the mobility needs of night-time workers or residents seeking to return home after late-night leisure. Coverage gaps were concentrated in outer and lower-income boroughs, and night bus frequency was insufficient to substitute for the closed Underground.

This finding translates directly to Shanghai. The Shanghai Metro's last departures cluster around 22:30–23:30 depending on line and direction, while the social norm for nightlife in Chinese cities — particularly bar districts, live music venues, and KTV — places the genuine peak of activity between 22:00 and 02:00. A resident living within walking distance of a bar in Jing'an may have excellent fifteen-minute entertainment access in the Moreno et al. sense, but if the last metro departs at 23:00 and the bar's peak hours begin at 22:00, the effective window for using that access without incurring substantial taxi cost is under sixty minutes. The McArthur et al. framework provides the intellectual grounding for treating this gap as an equity issue, not merely a logistical inconvenience: it reveals how the temporal structure of transit provision reproduces spatial inequality by restricting late-night access to those who can afford private transport.

The original contribution of this project is to apply the McArthur et al. insight to a Chinese megacity context using a data-driven grid approach, producing the first systematic map of the nightlife-transit coverage gap across all of Shanghai's 25,000 grid cells.

### 5. Shanghai Urban Vitality and the Inner Ring Gradient

Research on Shanghai's urban spatial structure consistently documents a strong centre-periphery gradient in vitality metrics. Liu et al. (2023), using multisource urban data including POI density, mobile phone activity, and nighttime light imagery, find that urban vitality in Shanghai presents a "strongly commercially driven spatial agglomeration pattern" during nighttime hours, with young residents' activity significantly more centrally concentrated than middle-aged or older cohorts. This is precisely the demographic most likely to use nightlife entertainment, and the finding predicts that Track B scores will exhibit a steep radial decay from the inner ring outward — a spatial pattern this project quantifies and maps at the grid-cell level.

The Five New Towns (Jiading, Songjiang, Qingpu, Fengxian, Nanhui) represent a deliberate policy counter-narrative: billions of yuan of infrastructure investment aimed at redistributing urban function to the periphery. Whether entertainment and nighttime transit have followed this investment — or whether residents of the new towns remain dependent on private transport to access cultural and leisure amenities — is the policy question this project is positioned to answer empirically.

## 2 · Generate 500 m × 500 m fishnet grid clipped to boundary

Strategy: snap grid origin to a clean multiple of 500 m, iterate all cells,  
keep only those whose **centroid** falls inside Shanghai.  
This gives clean, non-partial cells — the same philosophy used in the H3 notebook's `fill_hexagons()`.

```python
def make_fishnet(boundary_utm: gpd.GeoDataFrame, cell_size: float = 500.0) -> gpd.GeoDataFrame:
    """
    Generate a regular square fishnet clipped to the boundary polygon.

    Uses centroid-in-boundary test (same logic as H3 polyfill:
    only keep cells whose centre is inside), so every retained cell
    is assigned unambiguously — no partial or split cells.

    Parameters
    ----------
    boundary_utm : GeoDataFrame in a metric CRS (UTM)
    cell_size    : side length in metres (default 500)

    Returns
    -------
    GeoDataFrame of full square cells whose centroid is inside the boundary.
    """
    minx, miny, maxx, maxy = boundary_utm.total_bounds
    boundary_geom = boundary_utm.geometry.iloc[0]

    # snap to clean grid origin (floor to nearest cell_size multiple)
    minx = np.floor(minx / cell_size) * cell_size
    miny = np.floor(miny / cell_size) * cell_size

    xs = np.arange(minx, maxx + cell_size, cell_size)
    ys = np.arange(miny, maxy + cell_size, cell_size)

    cells = []
    for x in xs:
        for y in ys:
            cell = box(x, y, x + cell_size, y + cell_size)
            # centroid-in-boundary test → unambiguous assignment
            if boundary_geom.contains(cell.centroid):
                cells.append(cell)

    grid = gpd.GeoDataFrame(geometry=cells, crs=boundary_utm.crs)
    return grid


print("Generating 500 m fishnet … (≈ 1–2 min for full Shanghai)")
grid_utm = make_fishnet(shanghai_utm, cell_size=500)
print(f"\nGrid cells generated : {len(grid_utm):,}   (target ~25,000)")
```

## 3 · Extract centroid of each cell → `cell_id`, `lon`, `lat`

```python
# assign cell_id before anything else — never changes after this
grid_utm["cell_id"] = [f"SH_{i:06d}" for i in range(len(grid_utm))]

# compute centroids in UTM (accurate), then reproject to WGS84 for lon/lat
centroids_utm = grid_utm.copy()
centroids_utm["geometry"] = grid_utm.geometry.centroid

centroids_wgs = centroids_utm.to_crs(WGS84)
centroids_wgs["lon"] = centroids_wgs.geometry.x
centroids_wgs["lat"] = centroids_wgs.geometry.y

# start building master GDF — centroid geometry, WGS84
grid_master = centroids_wgs[["cell_id", "lon", "lat", "geometry"]].copy()
grid_master = grid_master.set_crs(WGS84)

print(f"Centroids : {len(grid_master):,}")
print(f"Lon range : {grid_master.lon.min():.4f} – {grid_master.lon.max():.4f}")
print(f"Lat range : {grid_master.lat.min():.4f} – {grid_master.lat.max():.4f}")
grid_master.head()
```

## 5 · Assign district and subdistrict via spatial join

Two-pass sjoin (same pattern as teacher's `spatial_join_districts` in cell [128]–[130]):  
1. district level → `sh-province-district.geojson`  
2. subdistrict level → `sh-towns.geojson`

```python
# ── Load district and town layers ─────────────────────────────────────────────
gdf_dist  = gpd.read_file(DISTRICT_PATH).to_crs(WGS84)
gdf_towns = gpd.read_file(TOWNS_PATH).to_crs(WGS84)

print("District columns :", gdf_dist.columns.tolist())
print("Towns columns    :", gdf_towns.columns.tolist())
gdf_dist.head(2)
```

```python
def pick_name_col(gdf, candidates=("name", "Name", "NAME", "district",
                                    "town", "subdistrict", "libcom", "NL_NAME_2")):
    """Auto-detect the name column from a GeoDataFrame."""
    for c in candidates:
        if c in gdf.columns:
            return c
    # fallback: first object column that isn't geometry
    for c in gdf.columns:
        if gdf[c].dtype == object and c != "geometry":
            return c
    raise ValueError(f"No name column found. Columns: {gdf.columns.tolist()}")

dist_name  = pick_name_col(gdf_dist)
towns_name = pick_name_col(gdf_towns)
print(f"Using '{dist_name}' for district, '{towns_name}' for subdistrict")
```

```python
# ── Pass 1: district ─────────────────────────────────────────────────────────
joined = gpd.sjoin(
    grid_master,
    gdf_dist[[dist_name, "geometry"]].rename(columns={dist_name: "district"}),
    how="left",
    predicate="within"
).drop(columns=["index_right"], errors="ignore")

# ── Pass 2: subdistrict (towns) ───────────────────────────────────────────────
joined = gpd.sjoin(
    joined,
    gdf_towns[[towns_name, "geometry"]].rename(columns={towns_name: "subdistrict"}),
    how="left",
    predicate="within"
).drop(columns=["index_right"], errors="ignore")

# drop boundary duplicates (a point exactly on a border can match two polygons)
joined = joined.drop_duplicates(subset="cell_id").reset_index(drop=True)
grid_master = joined.copy()

print(f"Cells with district    : {grid_master.district.notna().sum():,}")
print(f"Cells with subdistrict : {grid_master.subdistrict.notna().sum():,}")
print(f"Unique districts       : {grid_master.district.nunique()}")
print(f"Unique subdistricts    : {grid_master.subdistrict.nunique()}")
```

## 6 · Tag ring zone — inner / middle / outer / new_town

Shanghai has no hard administrative ring-road boundaries.  
Two approaches are provided:

**Option A (default — runs immediately):** radial distance from People's Square.  
| Zone | Radius (approx) |
|---|---|
| `inner_ring` | < 4 km |
| `middle_ring` | 4 – 8 km |
| `outer_ring` | 8 – 18 km |
| `new_town` | > 18 km |

**Option B (commented — use after drawing in QGIS):** load your own ring-road shapefile/GeoJSON drawn in QGIS and do a spatial join. Uncomment that block once your file is ready.

```python
# ── Option A: radial distance from People's Square ────────────────────────────
# This is the recommended approach when no ring-road shapefile is available.
# People's Square (人民广场) — geographic centre of central Shanghai
CENTRE_LON, CENTRE_LAT = 121.4737, 31.2304

# project to UTM for accurate planar distance
transformer = Transformer.from_crs(WGS84, UTM51N, always_xy=True)
cx, cy = transformer.transform(CENTRE_LON, CENTRE_LAT)
pts_x, pts_y = transformer.transform(
    grid_master["lon"].values,
    grid_master["lat"].values
)
dist_m = np.sqrt((pts_x - cx)**2 + (pts_y - cy)**2)
grid_master["dist_centre_m"] = dist_m

# thresholds (metres)
INNER_R  =  4_000
MIDDLE_R =  8_000
OUTER_R  = 18_000

def classify_ring(d):
    if d < INNER_R:
        return "inner_ring"
    elif d < MIDDLE_R:
        return "middle_ring"
    elif d < OUTER_R:
        return "outer_ring"
    else:
        return "new_town"

grid_master["ring_zone"] = pd.Series(dist_m).apply(classify_ring).values

print("Ring zone distribution (Option A — radial buffer):")
vc = grid_master.ring_zone.value_counts()
for zone, cnt in vc.items():
    pct = cnt / len(grid_master) * 100
    print(f"  {zone:<15}: {cnt:>6,}  ({pct:.1f}%)")
```

```python
RING_COLOURS = {
    "inner_ring" : "#e41a1c",
    "middle_ring": "#ff7f00",
    "outer_ring" : "#4daf4a",
    "new_town"   : "#377eb8",
}

sample = grid_final.sample(min(2000, len(grid_final)), random_state=42)

m = Map(
    location=[31.23, 121.47],
    zoom_start=10,
    tiles="CartoDB positron"
)

for _, row in sample.iterrows():
    folium.CircleMarker(
        location=[row["lat"], row["lon"]],
        radius=2,
        color=RING_COLOURS.get(row["ring_zone"], "grey"),
        fill=True,
        fill_opacity=0.75,
        weight=0,
        tooltip=(
            f"<b>{row['cell_id']}</b><br>"
            f"H3: {row['h3_index']}<br>"
            f"District: {row['district']}<br>"
            f"Subdistrict: {row['subdistrict']}<br>"
            f"Ring: {row['ring_zone']}"
        )
    ).add_to(m)

# mark People's Square
folium.Marker(
    location=[CENTRE_LAT, CENTRE_LON],
    tooltip="People's Square — ring centre",
    icon=folium.Icon(color="red", icon="star")
).add_to(m)

# legend
legend_html = """
<div style="position:fixed; bottom:30px; left:30px; z-index:9999;
            background:white; padding:12px 16px; border-radius:8px;
            border:1px solid #ccc; font-size:13px; line-height:2;">
  <b>Ring Zone</b><br>
  <span style='color:#e41a1c'>●</span> inner_ring  (&lt; 4 km)<br>
  <span style='color:#ff7f00'>●</span> middle_ring (4–8 km)<br>
  <span style='color:#4daf4a'>●</span> outer_ring  (8–18 km)<br>
  <span style='color:#377eb8'>●</span> new_town    (&gt; 18 km)
</div>
"""
m.get_root().html.add_child(folium.Element(legend_html))

MAP_PATH = os.path.join(OUTPUT_DIR, "grid_master_check.html")
m.save(MAP_PATH)
print(f"Map saved → {MAP_PATH}")
m
```
